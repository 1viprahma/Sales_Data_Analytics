# Cleaning

## Handling Data Types

```python
df_customers['CustomerID'] = df_customers['CustomerID'].str[1:].astype(int)
df_products['ProductID'] = df_products['ProductID'].str[1:].astype(int)
df_stores['StoreID'] = df_stores['StoreID'].str[1:].astype(int)
```

```python
df_transactions['TransactionID'] = df_transactions['TransactionID'].str[1:].astype(int)
df_transactions['CustomerID'] = df_transactions['CustomerID'].str[1:].astype(int)
df_transactions['ProductID'] = df_transactions['ProductID'].str[1:].astype(int)
df_transactions['StoreID'] = df_transactions['StoreID'].str[1:].astype(int)
```

```python
for name, data in dict.items():
    print(f"_____{name}______")
    print(data.dtypes)
```

## Creating Dimension Table For Date 

```python
df_dates = pd.DataFrame({'Date': sorted(df_transactions['Date'].unique())})
df_dates['DateID'] = range(1, len(df_dates) + 1)
df_dates = df_dates[['DateID', 'Date']]
```

```python
df_transactions = df_transactions.merge(df_dates, on='Date', how='left')
df_transactions.drop(columns=['Date'], inplace=True)
```

## Creating Dimension Table For Payment Method

```python
df_payment_methods = pd.DataFrame({'PaymentMethod': sorted(df_transactions['PaymentMethod'].unique())})
df_payment_methods['PaymentMethodID'] = range(1, len(df_payment_methods) + 1)
df_payment_methods = df_payment_methods[['PaymentMethodID', 'PaymentMethod']]
```

```python
df_transactions = df_transactions.merge(df_payment_methods, on='PaymentMethod', how='left')
df_transactions.drop(columns=['PaymentMethod'], inplace=True)
```

## Merging Table Transactions With Table Products

```python
df_transactions = df_transactions.merge(df_products[['ProductID', 'UnitPrice', 'CostPrice']], on='ProductID', how='left')
df_products.drop(columns=['UnitPrice', 'CostPrice'], inplace=True)
```

## Creating New Columns 

```python
df_transactions['PriceBeforeDiscount'] = (df_transactions['UnitPrice'] * df_transactions['Quantity']).round(2)

df_transactions['TotalPriceAfterDiscount'] = ((df_transactions['UnitPrice'] * df_transactions['Quantity']) * (1 - df_transactions['Discount'])).round(2)

df_transactions['TotalCost'] = (df_transactions['CostPrice'] * df_transactions['Quantity']).round(2)

df_transactions['Profit'] = (df_transactions['TotalPriceAfterDiscount'] - df_transactions['TotalCost']).round(2)
```

```python
df_dates['Year'] = df_dates['Date'].dt.year
df_dates['Month'] = df_dates['Date'].dt.month
df_dates['MonthName'] = df_dates['Date'].dt.strftime('%B')
df_dates['Quarter'] = 'Q' + df_dates['Date'].dt.quarter.astype(str)
df_dates['Day'] = df_dates['Date'].dt.day
df_dates['DayOfWeek'] = df_dates['Date'].dt.strftime('%A')
```

```python
df_customers['FullName'] = df_customers['FirstName'] + ' ' + df_customers['LastName']
df_customers.drop(columns=['FirstName', 'LastName'], inplace=True)

current_year = 2026
df_customers['Age'] = current_year - df_customers['BirthDate'].dt.year

display(df_customers.head())
```

```python
print("--- Transactions Table ---")
display(df_transactions.head())

print("--- Dates Table ---")
display(df_dates.head())

print("--- Payment Methods Table ---")
display(df_payment_methods.head())

print("--- Products Table ---")
display(df_products.head())
```

```python
path = r"E:\NTI\Project\cleaned_retail_sales.xlsx"

tables = {
    'Customers': df_customers,
    'Products': df_products,
    'Stores': df_stores,
    'Fact_Transactions': df_transactions,
    'Dates': df_dates,
    'PaymentMethods': df_payment_methods
}

with pd.ExcelWriter(path) as writer:
    for name, df in tables.items():
        df.to_excel(writer, sheet_name=name, index=False)
```
